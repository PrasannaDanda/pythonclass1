```python
person={'name':'chinni','age':20,'city':'vijayawada'}
print(person)
print("accessing and modifying the person age:")
person["age"]=20
print(person)
print("adding and removing items")
person['email']='vkomarapu@gmail.com'
print(person)
del person['city']
print(person)
print("All key &values")
print(person.key())
print(person.values())
print(person.items())
print(person.get("age"))



```
